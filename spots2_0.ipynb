```python
from ultralytics import YOLO
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import drive
import torch

class DarkSpotDetector:
    def __init__(self, model_path, conf_threshold=0.25, iou_threshold=0.45):
        self.model = YOLO(model_path)
        self.model.conf = conf_threshold
        self.model.iou = iou_threshold
        self.color = (0, 0, 255)  # Red color (BGR format)

    def preprocess_image(self, image):
        """Enhance image for better dark spot detection"""
        lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
        l, a, b = cv2.split(lab)

        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
        cl = clahe.apply(l)

        enhanced_lab = cv2.merge([cl, a, b])
        enhanced_bgr = cv2.cvtColor(enhanced_lab, cv2.COLOR_LAB2BGR)

        return enhanced_bgr

    def draw_arrow(self, img, target_point, offset_angle=45, arrow_length=50):
        """Draw an arrow pointing to the dark spot"""
        x, y = target_point

        # Calculate start point of arrow
        start_x = int(x + arrow_length * np.cos(np.radians(offset_angle)))
        start_y = int(y - arrow_length * np.sin(np.radians(offset_angle)))

        # Draw the arrow
        cv2.arrowedLine(img,
                       (start_x, start_y),
                       (int(x), int(y)),
                       self.color,
                       2,
                       tipLength=0.3)

        return start_x, start_y

    def draw_box(self, img, box):
        """Draw a semi-transparent box around the dark spot"""
        x, y, w, h = box
        x1 = int(x - w/2)
        y1 = int(y - h/2)
        x2 = int(x + w/2)
        y2 = int(y + h/2)

        # Draw semi-transparent fill
        overlay = img.copy()
        cv2.rectangle(overlay, (x1, y1), (x2, y2), self.color, -1)
        img = cv2.addWeighted(overlay, 0.2, img, 0.8, 0)

        # Draw border
        cv2.rectangle(img, (x1, y1), (x2, y2), self.color, 2)

        return img

    def detect_and_draw(self, image_path, min_size=20):
        """Detect dark spots and draw both boxes and arrows"""
        # Read and preprocess image
        original_img = cv2.imread(image_path)
        if original_img is None:
            raise ValueError(f"Could not read image at {image_path}")

        enhanced_img = self.preprocess_image(original_img)

        # Run detection
        results = self.model(enhanced_img)
        processed_img = enhanced_img.copy()

        # Store detected information
        detected_info = []  # Will store (center_point, box) pairs

        # Process each detection
        for result in results:
            if result.boxes is None:
                continue

            boxes = result.boxes.xywh.cpu().numpy()  # xcycwh format
            classes = result.boxes.cls.cpu().numpy()

            for box, cls in zip(boxes, classes):
                # Filter by size
                if box[2] < min_size or box[3] < min_size:
                    continue

                center_point = (int(box[0]), int(box[1]))
                detected_info.append((center_point, box))

        # If we found any dark spots, draw visualizations
        if detected_info:
            # Calculate label position
            top_y = min(point[1] for point, _ in detected_info) - 40
            avg_x = sum(point[0] for point, _ in detected_info) / len(detected_info)

            # Draw label
            label_text = "Darkspots"
            font = cv2.FONT_HERSHEY_SIMPLEX
            font_scale = 0.8
            thickness = 2

            # Get text size
            (text_width, text_height), _ = cv2.getTextSize(
                label_text, font, font_scale, thickness)

            # Calculate label position
            label_x = int(avg_x - text_width/2)
            label_y = max(top_y, text_height + 10)

            # Draw label background
            padding = 5
            cv2.rectangle(processed_img,
                         (label_x - padding, label_y - text_height - padding),
                         (label_x + text_width + padding, label_y + padding),
                         self.color, -1)

            # Draw label text
            cv2.putText(processed_img, label_text,
                       (label_x, label_y), font, font_scale,
                       (255, 255, 255), thickness)

            # Draw boxes and arrows for each detection
            for i, (point, box) in enumerate(detected_info):
                # Draw box
                processed_img = self.draw_box(processed_img, box)

                # Draw arrow with varied angle
                angle = 45 + (i * (90 / (len(detected_info) + 1)))
                self.draw_arrow(processed_img, point, offset_angle=angle)

        return processed_img, len(detected_info) > 0

def main():
    # Mount Google Drive
    drive.mount('/content/drive')

    # Initialize detector
    detector = DarkSpotDetector(
        model_path='/content/drive/My Drive/Trained_models/spots_model.pt',
        conf_threshold=0.25,
        iou_threshold=0.45
    )

    # Process image
    image_path = '/content/drive/My Drive/dark_spots/dark_spots/dark                                                                                                                                                                                                   .jpg'
    try:
        # Detect dark spots and draw on the image
        result_image, dark_spots_detected = detector.detect_and_draw(image_path)

        # Display the result image
        cv2_imshow(result_image)

        # Print detection result
        if dark_spots_detected:
            print("Dark spots detected!")
        else:
            print("No dark spots detected.")
    except Exception as e:
        print(f"Error processing image: {str(e)}")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def analyze_yolo_accuracy(results_df):
    """Analyze and visualize YOLO training accuracy metrics"""
    # Get the final epoch metrics
    final_metrics = results_df.iloc[-1]

    # Calculate final accuracy metrics (convert to percentages)
    final_accuracy = {
        'Precision': final_metrics['metrics/precision(B)'] * 100,
        'Recall': final_metrics['metrics/recall(B)'] * 100,
        'mAP50': final_metrics['metrics/mAP50(B)'] * 100,
        'mAP50-95': final_metrics['metrics/mAP50-95(B)'] * 100
    }

    # Create two subplots
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

    # Plot accuracy metrics
    epochs = range(1, len(results_df) + 1)
    ax1.plot(epochs, results_df['metrics/precision(B)'] * 100, label='Precision', marker='o')
    ax1.plot(epochs, results_df['metrics/recall(B)'] * 100, label='Recall', marker='o')
    ax1.set_xlabel('Epochs')
    ax1.set_ylabel('Percentage (%)')
    ax1.set_title('Precision and Recall over Training Epochs')
    ax1.grid(True)
    ax1.legend()

    # Plot mAP metrics
    ax2.plot(epochs, results_df['metrics/mAP50(B)'] * 100, label='mAP50', marker='o')
    ax2.plot(epochs, results_df['metrics/mAP50-95(B)'] * 100, label='mAP50-95', marker='o')
    ax2.set_xlabel('Epochs')
    ax2.set_ylabel('Percentage (%)')
    ax2.set_title('Mean Average Precision (mAP) over Training Epochs')
    ax2.grid(True)
    ax2.legend()

    plt.tight_layout()
    return final_accuracy, plt.gcf()

# Convert your data to DataFrame
data = pd.read_csv('/content/drive/MyDrive/Trained_models/spots_model/results.csv')  # Replace with your path

# Analyze results
final_metrics, plot = analyze_yolo_accuracy(data)

# Print final metrics
print("\nFinal Model Accuracy Metrics:")
for metric, value in final_metrics.items():
    print(f"{metric}: {value:.2f}%")

# Show plot
plt.show()
```
