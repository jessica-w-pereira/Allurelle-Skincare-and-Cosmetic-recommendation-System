```python
import cv2
import numpy as np
from ultralytics import YOLO

def detect_acne(image_path):
    model = YOLO("yolov8n.pt")
    results = model.predict(image_path, conf=0.5)
    return results

def draw_arrow(img, target_point, offset_angle=45, arrow_length=50):
    """Draw an arrow pointing to the acne spot"""
    x, y = target_point
    
    # Calculate start point of arrow
    start_x = int(x + arrow_length * np.cos(np.radians(offset_angle)))
    start_y = int(y - arrow_length * np.sin(np.radians(offset_angle)))
    
    # Draw the arrow
    cv2.arrowedLine(img, 
                   (start_x, start_y),
                   (int(x), int(y)),
                   (0, 0, 255),  # Red color
                   2,            # Thickness
                   tipLength=0.3)
    
    return start_x, start_y

def draw_boxes_and_arrows(image_path, results, scale=0.5):
    img = cv2.imread(image_path)
    detected_info = []  # Store (center_point, box_coords) pairs
    
    for result in results:
        boxes = result.boxes
        for box in boxes:
            x1, y1, x2, y2 = box.xyxy[0]
            
            # Calculate center point
            center_x = int((x1 + x2) / 2)
            center_y = int((y1 + y2) / 2)
            
            # Calculate scaled box dimensions
            width = (x2 - x1) * scale
            height = (y2 - y1) * scale
            
            # Calculate new box coordinates
            new_x1 = int(center_x - width / 2)
            new_y1 = int(center_y - height / 2)
            new_x2 = int(center_x + width / 2)
            new_y2 = int(center_y + height / 2)
            
            # Store detection info
            detected_info.append(((center_x, center_y), (new_x1, new_y1, new_x2, new_y2)))
    
    # If we found any spots, draw visualizations
    if detected_info:
        # Calculate label position (above all detections)
        top_y = min(coords[1] for (_, coords) in detected_info) - 40
        center_points = [point for point, _ in detected_info]
        avg_x = sum(point[0] for point in center_points) / len(center_points)
        
        # Draw single label for all detections
        label_text = "Acne"
        font = cv2.FONT_HERSHEY_SIMPLEX
        font_scale = 1.5
        thickness = 2
        box_color = (0, 0, 255)  # Red color
        
        # Get text size for centering
        (text_width, text_height), _ = cv2.getTextSize(
            label_text, font, font_scale, thickness)
        
        # Calculate label position
        label_x = int(avg_x - text_width/2)
        label_y = max(int(top_y), text_height + 10)
        
        # Draw label background
        padding = 5
        cv2.rectangle(img,
                     (label_x - padding, label_y - text_height - padding),
                     (label_x + text_width + padding, label_y + padding),
                     box_color, -1)
        
        # Draw label text
        cv2.putText(img, label_text,
                   (label_x, label_y), font, font_scale,
                   (255, 255, 255), thickness)
        
        # Draw boxes and arrows for each detection
        for i, (center_point, box_coords) in enumerate(detected_info):
            # Draw semi-transparent box
            overlay = img.copy()
            cv2.rectangle(img, 
                         (box_coords[0], box_coords[1]), 
                         (box_coords[2], box_coords[3]), 
                         box_color, 2)
            cv2.rectangle(overlay, 
                         (box_coords[0], box_coords[1]), 
                         (box_coords[2], box_coords[3]), 
                         box_color, -1)
            img = cv2.addWeighted(overlay, 0.2, img, 0.8, 0)
            
            
    
    # Display the image
    cv2.namedWindow("Image", cv2.WINDOW_NORMAL)
    cv2.resizeWindow("Image", 600, 600)
    cv2.imshow("Image", img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

# Example usage
image_path = "C:/Users/Anjelica/OneDrive/Desktop/purvi_acne.jpg"
results = detect_acne(image_path)
draw_boxes_and_arrows(image_path, results, scale=0.5)
```

```python
metrics = model.metrics  # Get metrics directly from model
print("Model Accuracy:")
print(f"Mean mAP50-95: {metrics.box.map * 100:.2f}%")  # mean mAP for all classes
print(f"Mean mAP50: {metrics.box.map50 * 100:.2f}%")   # mAP at 0.5 IOU
print(f"Mean Precision: {metrics.box.mp * 100:.2f}%")   # Using mp() for mean precision
print(f"Mean Recall: {metrics.box.mr * 100:.2f}%")      # Using mr() for mean recall
```
